## Fixing the DATASETS renaming and peprare to combine

```python
df_old.rename(columns={'team_x': 'team', 'season_x': 'season'}, inplace=True)
df_25['season'] = '2024-25'
df_current['season'] = '2025-26'
df_old.columns
```

## keeping only the must have columns

```python
common_columns = ['name', 'position', 'team', 'season', 'GW',
                 'minutes', 'total_points', 'value', 'selected',
                 'goals_scored', 'assists', 'clean_sheets', 'goals_conceded',
                 'own_goals', 'penalties_saved', 'penalties_missed', 'yellow_cards', 'red_cards',
                  'saves', 'bonus', 'bps', 'influence', 'creativity', 'threat', 'ict_index',
                  'transfers_balance', 'transfers_in', 'transfers_out',
                  'team_h_score', 'team_a_score', 'was_home', 'opponent_team', 'fixture', 'round', 'element'
                 ]
```

### Filtering and merging

```python
df_part1 = df_old[common_columns]
df_part2 = df_25[common_columns]
df_part3 = df_current[common_columns]
final_df = pd.concat([df_part1, df_part2, df_part3], axis=0, ignore_index=True)
```

### Filtering out the old seasons

```python
seasons_to_keep = ['2021-22', '2022-23', '2023-24', '2024-25', '2025-26']
final_df = final_df[final_df['season'].isin(seasons_to_keep)]
```

### exporting and saving the new file 

```python
final_df.to_csv('../Data/Cleaned/final_fpl_dataset.csv', index=False)
print(f"Sucess! {len(final_df)} rows.")
print("columns:", list(final_df.columns))
```
